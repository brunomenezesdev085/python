```python
preco = 50 #valor em real
desconto = 10 #desconto em %

novo_preco = preco - (preco * desconto / 100)
print(f'O novo valor com desconto é de R$ {novo_preco}')
```

```python
preco = float(input('Digite o valor do produto:'))
desconto = float(input('Digite o desconto do produto:'))

novo_preco = preco - (preco * desconto / 100)
print(f'O novo valor com desconto é de R$ {novo_preco}')
```
