```python
Quantidade_porcoes = float(input('Quantas porções voce tem do produto?'))
porcoes_por_dia = float(input('Voce usa quantas porcoes por dia?'))

dias = (Quantidade_porcoes/porcoes_por_dia)

print(f'O produto ira durar {dias} dias com essas quantidades')
```
